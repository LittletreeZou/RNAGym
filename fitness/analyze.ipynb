```python
import pandas as pd
import numpy as np
import os
```

```python
ref_file = "/lustre/scratch/shared-folders/bio_project/shuxian/gbft/mg/rna_202507/RNAGym/fitness/reference_sheet_final.csv"
metric_dir = "/lustre/scratch/shared-folders/bio_project/shuxian/gbft/mg/rna_202507/RNAGym/fitness_prediction/metric"

ref = pd.read_csv(ref_file)
ref.head()
```

```python
models = ["aido_rna_650m_cds", "aido_rna_650m"] #["aido_rna_1b600m_cds", "aido_rna_1b600m"]
methods = ["wt-marginals"]  #"masked-marginals",

for model in models:
    for method in methods:
        print(f"{model}-{method}")
        metric_path = os.path.join(metric_dir, f"{model}-{method}.csv")
        df = pd.read_csv(metric_path)
        merged = pd.merge(ref, df, left_on="DMS_ID", right_on="dms_id", how="left")
        print("RNA TYPE")
        mean_df = merged.groupby("RNA_TYPE")["spearman"].mean().reset_index()
        print(mean_df)
        mean_df = merged.groupby("RNA_TYPE")["auc"].mean().reset_index()
        print(mean_df)
        mean_df = merged.groupby("RNA_TYPE")["mcc"].mean().reset_index()
        print(mean_df)
        print("mRNA coding species")
        mean_df = merged[merged["RNA_TYPE"] == "mRNA-coding"].groupby("ASSAY_DESCRIPTION")["spearman"].mean().reset_index()
        print(mean_df)
        print("-------------")
```
